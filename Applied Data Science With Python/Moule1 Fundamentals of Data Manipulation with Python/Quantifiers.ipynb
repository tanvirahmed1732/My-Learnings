```python
import re
```

```python
grades="ABCDABBCADAAACB"
```

```python
re.findall('A{2,10}',grades)  # Find all occurrences of 'A' repeated 2 to 10 times in the grades string
#the curly braces {} are used to specify the number of times a character or group should be repeated.
```

```python
re.findall('A{1,1}A{1,1}',grades)  # Find all occurrences of 'A' repeated exactly 2 times in the grades string
#this is same as 'AA'
#the first number inside the curly braces is the minimum number of times the character should be repeated,
#and the second number is the maximum number of times the character can be repeated.
```

```python
#if we don't include a quantifier, it means exactly one occurrence of the preceding element. the default is {1,1}
re.findall('AA',grades)  # Find all occurrences of 'AA' in the grades string
```

```python
#if we include 1 numnber inside the curly braces, it means exactly that number of occurrences. it consider to be both min and max.
re.findall('A{3}',grades)  # Find all occurrences of 'A' repeated exactly 3 times in the grades string
```

```python
#using this we can find a decreasing trend of grades
re.findall('A{1,}B{1,}C{1,}D{1,}',grades)  # Find occurrences of decreasing grade patterns in the grades string
#this will find A followed by B followed by C followed by D with at least one occurrence of each.
#the quantifier {1,} means one or more occurrences of the preceding element.
re.findall('A{1,}B{1,}C{1,}',grades) # Find occurrences of decreasing grade patterns ending with C in the grades string
```
