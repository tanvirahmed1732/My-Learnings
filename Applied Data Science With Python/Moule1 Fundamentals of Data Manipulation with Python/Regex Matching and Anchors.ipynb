```python
import re
```

```python
#there are several main processing functions in re module like search(), match(), findall(), etc. the first match() function 
# thats checks for a match only at the beginning of the string. The search() function checks for a match anywhere in the string.
text="This is a good day"
re.search("good",text) # Search for the word "good" in the text
if re.search("good",text):
    print("Match found!")
else:
    print("No match found.")

"good" in text  # Check if "good" is in the text using the 'in' operator
```

```python
text="amy works delightly. amy gets good grades. our student amy is successful."
re.split("amy", text)  # Split the text at each occurrence of "amy"
```

```python
#here split() func returns a empty string at the start because "amy" is at the beginning of the text.
#if we want to avoid that we can use re.findall() function to find all occurrences of a pattern in a string.
re.findall("amy", text)  # Find all occurrences of "amy" in the text
```

```python
#let's see how anchors work in regex
# ^ asserts position at start of a line
# $ asserts position at end of a line
# if we put ^ at the start of our pattern it will match only if the pattern is at the start of the string
# similarly if we put $ at the end of our pattern it will match only if the pattern is at the end of the string
re.search("^amy", text)  # Check if the text starts with "amy"
re.search("successful.$", text)  # Check if the text ends with "successful."
```
