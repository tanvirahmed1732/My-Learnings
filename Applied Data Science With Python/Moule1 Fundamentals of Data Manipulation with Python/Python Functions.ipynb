```python
def add_numbers(x,y): #func to add 2 numbers
    return x+y

add_numbers(x,y)
```

```python
def add_3numbers(x, y, z=None):
    """Return sum of two or three numbers."""
    if z is None:
        return x + y
    return x + y + z

# Example calls
print(add_3numbers(x, y))
print(add_3numbers(1, 2, 3))
```
