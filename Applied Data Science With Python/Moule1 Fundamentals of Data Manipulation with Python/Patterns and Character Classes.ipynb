```python
import re
```

```python
grades="ABCDABBCADAAACB"
re.findall('B',grades)  # Find all occurrences of 'B' in the grades string
```

```python
print(re.findall('AB',grades))  # Find all occurrences of the sequence 'AB' in the grades string
#if we want to find either A or B we can use character classes
re.findall('[AB]',grades)  # Find all occurrences of 'A' or 'B' in the grades string
```

```python
#this is called the set operator, it matches any one of the characters inside the square brackets.
#we can also use ranges inside the square brackets. for example [A-C] will match any character from A to C.
#student recieve A followed by B or C
re.findall('A[B-C]',grades)  # Find all occurrences of 'A' followed by 'B' or 'C' in the grades string
```

```python
#pipe operator '|' is used to match either the pattern before or the pattern after the operator.
#for example 'A|B' will match either A or B.
re.findall('AB|AC',grades)  # Find all occurrences of 'AB' or 'AC' in the grades string
```

```python
re.findall('[^A]',grades)  # Find all occurrences of characters that are not 'A' in the grades string
#the caret (^) inside the square brackets negates the character class, so it matches any character that is not in the class.
```

```python
re.findall('^[^A]',grades)  # Find the first character in the grades string that is not 'A'
#the caret (^) outside the square brackets matches the start of the string.
```
