```python
numbers=[1,2,3]
pd.Series(numbers) 
```

```python
students = ['Alice', 'Bob', 'Charlie', None]
pd.Series(students)
```

```python
numbers=[1,2,3,None]
pd.Series(numbers) # Notice how Pandas handles the None value, it converts the entire Series to float type to accommodate the NaN value.
#underneath pandas represents None as NaN (Not a Number) which is a special floating-point value.
#None is not equivalent to NaN; None is a Python object, while NaN is a special floating-point value defined by the IEEE floating-point standard.
```

```python
import numpy as np
np.NaN==None # This will return False because NaN is not equal to anything, including None.
```

```python
np.NaN==np.NaN # This will return False because NaN is not equal to anything, including itself.
```

```python
#instead we need to use the function isnan to check for NaN values
np.isnan(np.NaN) # This will return True because the value is NaN.
#so keep in mind NaN is similar to None but it's a numaric value and treated differently for efficiency reasons.
```
