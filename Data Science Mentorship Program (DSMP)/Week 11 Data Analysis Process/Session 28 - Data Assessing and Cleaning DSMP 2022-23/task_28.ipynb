### Data Quality Dimensions

- Completeness -> is data missing?
- Validity -> is data invalid -> negative height -> duplicate patient id
- Accuracy -> data is valid but not accurate -> weight -> 1kg
- Consistency -> both valid and accurate but written differently -> New Youk and NY
### Order of severity

Completeness <- Validity <- Accuracy <- Consistency
### Data Cleaning Order

1. Quality -> Completeness

### We need to seperate the stars column to stars and ratings using regex

```python
# Convert current stars text into separate numeric columns:
# e.g. '4.5 out of 5 stars38 ratings' -> stars_value=4.5, ratings_value=38
# Non-rated rows become stars_value=0 and ratings_value=0.


mask_not_rated = df['stars'].str.contains('Not rated yet', case=False, na=False)

# initialize with zeros
df['stars_value'] = 0.0
df['ratings_value'] = 0

# extract from valid rows (if any)
extracted = df.loc[~mask_not_rated, 'stars'].str.extract(
    r"(?P<stars_value>[0-9]+(?:\.[0-9]+)?)\s*out of\s*5\s*stars\s*(?P<ratings_value>[0-9]+)\s*ratings",
    expand=True
)

# only assign when extraction succeeds
valid = extracted['stars_value'].notna() & extracted['ratings_value'].notna()
if valid.any():
    df.loc[~mask_not_rated, 'stars_value'] = extracted.loc[valid, 'stars_value'].astype(float)
    df.loc[~mask_not_rated, 'ratings_value'] = extracted.loc[valid, 'ratings_value'].astype(int)

# clean final dtypes
df['stars_value'] = pd.to_numeric(df['stars_value'], errors='coerce').fillna(0.0).astype(float)
df['ratings_value'] = pd.to_numeric(df['ratings_value'], errors='coerce').fillna(0).astype(int)

# check result
print(df[['stars', 'stars_value', 'ratings_value']].head())
```

```python
df.drop(columns=['stars'],inplace=True)
df.head()
```

### Fixing the author names by removing the Writtenby text from each row

```python
df['author']=df['author'].str.split(":").str[1]
df.head()
```

### Fixing the narrator names by removing the Narratedby text from each row

```python
df['narrator']=df['narrator'].str.split(':').str[1]
# df.drop(columns=['narrattor'],inplace=True)
df.head()
```

### Replace the time to minutes

```python
# Standardize and extract minutes from time column
# Handles formats like '2 hrs and 20 mins', '45 min', '1h 15m', '90', etc.

time_s = df['time'].astype(str).str.lower().fillna('')

# Normalize common tokens
time_s = time_s.str.replace(r'hrs?', 'hr', regex=True)
time_s = time_s.str.replace(r'hours?', 'hr', regex=True)
time_s = time_s.str.replace(r'mins?', 'min', regex=True)
time_s = time_s.str.replace(r'minutes?', 'min', regex=True)

time_s = time_s.str.replace(r'\band\b', ' ', regex=True)
time_s = time_s.str.replace(r'(?P<num>\d+)\s*h\b', r'\g<num> hr', regex=True)
time_s = time_s.str.replace(r'(?P<num>\d+)\s*m\b', r'\g<num> min', regex=True)

parts = time_s.str.extract(
    r'^\s*(?:(?P<h>\d+)\s*hr\s*)?(?:(?P<m>\d+)\s*min\s*)?$'
)

parts = parts.fillna(0).astype(int)

df['time_min'] = parts['h'] * 60 + parts['m']

# fallback 0 for any invalid/missing patterns
df['time_min'] = df['time_min'].fillna(0).astype(int)

# verify
df[['time', 'time_min']].head()
```
