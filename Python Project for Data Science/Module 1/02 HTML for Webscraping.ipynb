# HTML Basics for Web Scraping – Study Notes

## Overview
HTML (Hypertext Markup Language) is the standard language used to structure content on the web. Many valuable datasets such as real estate prices, sports statistics, and answers to coding questions are embedded in web pages as HTML. By understanding HTML structure, we can use Python to extract this data for analysis and automation.

Websites like Wikipedia act as large repositories of information, making HTML knowledge essential for effective web scraping.

---

## Learning Objectives
By understanding HTML, you can:
- Read and interpret the HTML of a basic web page
- Understand the structure and components of an HTML tag
- Navigate HTML document trees
- Extract data from HTML tables

---

## Structure of an HTML Web Page
An HTML document is composed of text and **tags**, which are written inside angle brackets (`< >`). Tags tell the browser how to display content.

### Key Components
- `<!DOCTYPE html>`  
  Declares that the document is an HTML document.
- `<html>`  
  The root element of the HTML page.
- `<head>`  
  Contains meta-information about the page (not usually visible).
- `<body>`  
  Contains the visible content of the web page.  
  This is usually where the data we want to scrape is located.

### Common Tags
- `<h3>`  
  A level-3 heading. Used for titles such as player names.
- `<p>`  
  Paragraph tag. Often used for descriptive text such as salaries.

**Example use case:**  
Player names may be stored inside `<h3>` tags, while salaries may be stored inside `<p>` tags. The data is enclosed between an opening tag (`<h3>`) and a closing tag (`</h3>`).

---

## Composition of an HTML Tag
An HTML tag consists of:
- **Tag name** (e.g., `a`, `p`, `h3`)
- **Opening tag** (start tag)
- **Closing tag** (end tag, preceded by `/`)
- **Content** (text shown on the page)
- **Attributes** (additional information)

### Example: Anchor (`<a>`) Tag
An anchor tag creates a hyperlink.

- Tag name: `a`
- Attribute name: `href`
- Attribute value: URL of the destination
- Content: text shown on the page (e.g., "IBM")

**Conceptual analogy:**  
Think of a tag name as a Python class and each tag as an instance of that class.

## HTML Document Tree (DOM)
An HTML document can be represented as a **tree structure**, often called the DOM (Document Object Model).

### Tree Relationships
- **Parent:** A tag that contains another tag
- **Child:** A tag nested inside another tag
- **Sibling:** Tags on the same level
- **Descendant:** Any tag nested inside another tag (not necessarily a direct child)

### Example Relationships
- `<html>` is the parent of `<head>` and `<body>`
- `<head>` and `<body>` are siblings
- `<title>` is a child of `<head>`
- Headings (`<h1>`, `<h3>`) and paragraphs (`<p>`) are children of `<body>`
- A `<b>` (bold) tag inside a heading is a child of that heading

---

## HTML Tables
HTML tables are commonly used to display structured data.

### Table Structure
- `<table>`  
  Defines the table
- `<tr>`  
  Defines a table row
- `<th>`  
  Defines a header cell (usually the first row)
- `<td>`  
  Defines a standard table cell

Each row (`<tr>`) contains multiple cells (`<td>`). This structure makes tables ideal for scraping tabular data like player names, salaries, or statistics.
