# Web Scraping with Python and BeautifulSoup – Study Notes

## What is Web Scraping?
Web scraping is the process of **automatically extracting information from websites**. Instead of manually copying and pasting data into spreadsheets, web scraping allows us to collect large amounts of data in **minutes rather than hours** using code.

It is especially useful when:
- You need to analyze hundreds or thousands of data points
- Data is spread across web pages
- Manual collection would be slow and error-prone

---

## Learning Objectives
After understanding this topic, you should be able to:
- Define web scraping
- Understand the role of BeautifulSoup objects
- Use the `find_all()` method
- Scrape real data from a website using Python

---

## Tools Used for Web Scraping
To scrape websites with Python, we mainly use two libraries:

- **Requests**  
  Used to download the HTML content of a webpage.
- **BeautifulSoup (bs4)**  
  Used to parse HTML and extract data from it.

---

## Basic Web Scraping Workflow
1. Send a request to a webpage using `requests`
2. Get the HTML content as text
3. Parse the HTML using BeautifulSoup
4. Navigate the HTML tree to extract required data

---


```python
from bs4 import BeautifulSoup

html = "<!DOCTYPE html><html><head><title>Page Title</title></head><body><h3><b id='boldest'>Lebron James</b></h3><p> Salary: $ 92,000,000 </p><h3> Stephen Curry</h3><p> Salary: $85,000,000 </p><h3> Kevin Durant </h3><p> Salary: $73,200,000</p></body></html>"

soup = BeautifulSoup(html, 'html5lib')
```

```python
tag_object=soup.title # here we are accessing the title tag
tag_object
```

```python
tag_object=soup.h3 # getting the first h3 tag
tag_object
```

```python
tag_child=tag_object.b # accessing the child tag <b> of h3 tag
tag_child
```

```python
parent_tag=tag_child.parent # accessing the parent tag of <b> tag
parent_tag
```

```python
sibling1=tag_object.next_sibling # accessing the next sibling of h3 tag
sibling1
```

```python
sibling2=sibling1.next_sibling # accessing the next sibling of the previous sibling
sibling2
```

```python
tag_child.attrs # we can access the attributes of a tag using .attrs property
```

```python
tag_child.string # we can access the string inside a tag using .string property
```

```python
html = "<table><tr><td>Pizza Place</td><td>Orders</td><td>Slices </td></tr><tr><td>Domino's Pizza</td><td>10</td><td>100</td></tr><tr><td>Little Caesars</td><td>12</td><td>144</td></table>"

table = BeautifulSoup(html, 'html5lib')
```

```python
table_rows=table.find_all('tr') # finding all the rows in the table
table_rows
```

```python
import requests
page=requests.get("https://www.zazzle.com/s/dinosaur+birthday+invitations?st=orderitemcount_month").text
soup=BeautifulSoup(page,"html.parser")
links=soup.find_all('a')
for link in links:
    fulllink=link.get('href')
    title=link.get('aria-label')
    print(fulllink)
    if title:
        print(title.replace("Product:", "").strip())

# here we are getting the links of all the products on the first page of dinosaur birthday invitations and printing them out along with their titles.
```
