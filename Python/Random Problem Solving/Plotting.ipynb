```python
import matplotlib.pyplot as plt
```

```python
nvals=[]
linear=[]
quadratic=[]   
cubic=[]
exponential=[]
for n in range(0,30):
    nvals.append(n)
    linear.append(n)
    quadratic.append(n**2)
    cubic.append(n**3)
    exponential.append(1.5**n)
```

```python
plt.plot(nvals, linear) #we can plot a graph like this
# plt.xlim(0,30)
# plt.ylim(0,30)
# plt.xticks(range(0,31,1))
```

```python
plt.figure('expo')
plt.plot(nvals, exponential)
plt.figure('lin')
plt.plot(nvals, linear)
plt.figure('quad')
plt.plot(nvals, quadratic) 
plt.figure('cubic')
plt.plot(nvals, cubic)
newExpo=[]
for i in range(30):
    newExpo.append(1.6**i) #taking another array
plt.figure('expo')
plt.plot(nvals, newExpo) #plotting it on the same graph as previous exponential
# plt.show() #to show all the graphs at once
```

```python
months=range(1,13)
boston =[28, 32, 39, 48, 60, 72, 80, 78, 70, 58, 45, 32]
plt.plot(months, boston, 'b-', label="Boston") #b- means blue line
pheonix =[55, 58, 65, 75, 85, 95, 100, 98, 90, 75, 65, 55]
plt.plot(months, pheonix, 'r--', label="Pheonix") #r-- means red dashed line
msp =[20, 25, 30, 40, 55, 70, 75, 73, 65, 50, 35, 25]  
plt.plot(months, msp, 'g-.', label="Minneapolis") #green dash-dot line
plt.legend(loc='best', fontsize=20) #to show which line is for which city with fontsize 20
```

Line Style
 - solid line
 -- dashed line
 -. dash dot line
 : dotted line
Color Options (plus many more)
 b blue
 g green
 r red
 c cyan
 m magenta
 y yellow
 k black
w white
Marker Options (plus many more)
. point
o circle
v triangle down
^ triangle up
* star

```python
months=range(1,13)
boston =[28, 32, 39, 48, 60, 72, 80, 78, 70, 58, 45, 32]
plt.plot(months, boston, '.b-', label="Boston") #b- means blue line with dot markers
pheonix =[55, 58, 65, 75, 85, 95, 100, 98, 90, 75, 65, 55]
plt.plot(months, pheonix, 'or--', label="Pheonix") #r-- means red dashed line with circle markers
msp =[20, 25, 30, 40, 55, 70, 75, 73, 65, 50, 35, 25]  
plt.plot(months, msp, '*g-.', label="Minneapolis") #green dash-dot line with star markers
plt.legend(loc='best', fontsize=20) #to show which line is for which city with fontsize 20
```
